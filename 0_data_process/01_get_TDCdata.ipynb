```python
from pathlib import Path

RAW = Path(r"../data/in_vitro_data")
RAW.mkdir(parents=True, exist_ok=True)

print("Save paths:")
print(RAW)
```

```python
import pandas as pd
from tdc.single_pred import ADME

datasets = {
    "CL_hepatocyte": {
        "tdc_name": "Clearance_Hepatocyte_AZ",
        "file": "tdc_cl_hep.csv",
    },
    "CL_microsome": {
        "tdc_name": "Clearance_Microsome_AZ",
        "file": "tdc_cl_mic.csv",
    },
    "half_life": {
        "tdc_name": "Half_Life_Obach",
        "file": "tdc_half_life.csv",
    },
    "ppb": {
        "tdc_name": "PPBR_AZ",
        "file": "tdc_PPBR_AZ.csv",
    },
}

tdc_data = {}

for key, info in datasets.items():

    print("\n" + "=" * 70)
    print(key.upper())
    print("=" * 70)
    print("TDC dataset:", info["tdc_name"])

    try:
        data = ADME(name=info["tdc_name"])
        df = data.get_data()

        print(f"Downloaded: {len(df):,} records")
        print("Columns:", df.columns.tolist())

        # Save
        output_path = RAW / info["file"]
        df.to_csv(output_path, index=False)

        print("Saved:", output_path)

        # Save so it can be used later in the notebook
        tdc_data[key] = df.copy()

    except Exception as e:
        print("ERROR:", repr(e))
```
