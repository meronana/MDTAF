```python
print("Step 1: Collecting SMILES from all CV folds...\n")

def get_smiles_column(df):
    """Find the SMILES column"""
    if 'smiles' in df.columns:
        return 'smiles'
    elif 'Drug' in df.columns:
        return 'Drug'
    else:
        return None

smiles_set = set()
file_count = 0

for fold_idx in range(N_FOLDS):
    fold_dir = os.path.join(cv_dir, f'fold_{fold_idx}')
    print(f"  Fold {fold_idx}:")

    for prop in PROPERTIES:
        # SOURCE
        src_path = os.path.join(fold_dir, f'{prop}_source.csv')
        if os.path.exists(src_path):
            df = pd.read_csv(src_path)
            smiles_col = get_smiles_column(df)
            if smiles_col:
                smiles_list = df[smiles_col].astype(str).tolist()
                smiles_set.update(smiles_list)
                print(f"    {prop}_source: {len(smiles_list)} SMILES ({smiles_col})")
                file_count += 1

        # TARGET train/test
        for split in ['train', 'test']:
            tgt_path = os.path.join(fold_dir, f'{prop}_target_{split}.csv')
            if os.path.exists(tgt_path):
                df = pd.read_csv(tgt_path)
                smiles_col = get_smiles_column(df)
                if smiles_col:
                    smiles_list = df[smiles_col].astype(str).tolist()
                    smiles_set.update(smiles_list)
                    print(f"    {prop}_target_{split}: {len(smiles_list)} SMILES ({smiles_col})")
                    file_count += 1

all_smiles = sorted(smiles_set)
print(f"\n✓ Total unique SMILES: {len(all_smiles)}")
print(f"✓ Files processed: {file_count}\n")
```

## Step 2: Load the Graphormer encoder

```python
print("Step 2: Loading Graphormer encoder...\n")

import sys
ROOT = r'..'
sys.path.insert(0, ROOT)

try:
    from src.graphormer_encoder import embed, load_encoder
    from src.graphormer_preprocess import collate, preprocess_smiles
    import torch

    DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
    print(f"✓ Imports successful")
    print(f"✓ Device: {DEVICE}\n")

    print("Loading encoder...")
    enc, cfg = load_encoder(DEVICE)
    print(f"✓ Encoder loaded\n")

except ImportError as e:
    print(f"✗ Import error: {e}")
    print("Note: Make sure src/graphormer_encoder.py exists")
    enc = None
```

## Step 3: Extract embeddings

```python
if enc is not None:
    print("Step 3: Extracting embeddings...\n")
    print(f"Processing {len(all_smiles)} molecules\n")

    MAX_NODES = 128
    BATCH_SIZE = 16

    keys = []
    vecs = []
    buf_items = []
    buf_keys = []
    n_fail = 0
    n_processed = 0

    start_time = time.time()

    def flush():
        global buf_items, buf_keys, n_processed, keys, vecs

        if not buf_items:
            return

        try:
            batch = collate(
                buf_items,
                spatial_pos_max=cfg.spatial_pos_max
            )

            emb = embed(enc, batch, DEVICE)

            # Tensor -> NumPy
            if isinstance(emb, torch.Tensor):
                emb = emb.detach().cpu().numpy()

            emb = emb.astype("float32")

            keys.extend(buf_keys)
            vecs.append(emb)
            n_processed += len(buf_items)

        except Exception as e:
            print(f"  Error in batch processing: {e}")

        finally:
            buf_items.clear()
            buf_keys.clear()

    for i, smi in enumerate(all_smiles):
        try:
            item = preprocess_smiles(smi)

            if item is None:
                n_fail += 1
                continue

            if item["num_nodes"] > MAX_NODES:
                n_fail += 1
                continue

            buf_items.append(item)
            buf_keys.append(smi)

        except Exception as e:
            n_fail += 1
            continue

        # Flush buffer
        if len(buf_items) >= BATCH_SIZE:
            flush()

        # Progress
        if (i + 1) % 1000 == 0:
            elapsed = time.time() - start_time

            print(
                f"  [{i+1:6d}/{len(all_smiles):6d}] "
                f"Embedded: {n_processed:5d} | "
                f"Failed: {n_fail:4d} | "
                f"Elapsed: {elapsed:.1f}s"
            )

    # Final flush
    flush()

    elapsed = time.time() - start_time

    # Concatenate
    if vecs:
        X = np.concatenate(vecs, axis=0).astype("float32")
    else:
        X = np.zeros((0, 768), dtype="float32")

    print("\n✓ Embedding extraction complete")
    print(f"  Total processed: {len(keys)}")
    print(f"  Total failed: {n_fail}")

    total = len(keys) + n_fail
    success_rate = len(keys) / total * 100 if total > 0 else 0.0

    print(f"  Success rate: {success_rate:.1f}%")
    print(f"  Elapsed time: {elapsed:.1f}s")
    print(f"  Embedding shape: {X.shape}\n")

else:
    print("Skipping embedding extraction (encoder not loaded)")
    keys = None
    X = None
```

## Step 4: Save

```python
if keys is not None and X is not None:
    print("Step 4: Saving embeddings...\n")

    output_path = os.path.join(feat_dir, 'graphormer_cv_embeddings.npz')

    print(f"Saving to: {output_path}")
    np.savez_compressed(
        output_path,
        keys=np.array(keys),
        X=X
    )

    file_size = os.path.getsize(output_path) / (1024**2)  # MB
    print(f"\n✓ Saved successfully")
    print(f"  File size: {file_size:.1f} MB")
    print(f"  Embeddings: {len(keys)}")
    print(f"  Dimensions: {X.shape[1]}\n")

else:
    print("Skipping save (embeddings not extracted)")
```
